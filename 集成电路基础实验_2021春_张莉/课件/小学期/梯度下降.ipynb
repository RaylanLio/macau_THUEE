# 设置噪声和原始数据

```python
nx = np.random.rand(10)
ny = np.random.rand(10)
plt.plot(nx,ny)
plt.grid('-')
```

```python
pre_x = np.array([i for i in range(10)])
pre_y = 0.5*pre_x+20
plt.plot(pre_x,pre_y)
plt.grid('-')
```

```python
x = pre_x + nx
y = pre_y + ny
plt.plot(x,y)
```
